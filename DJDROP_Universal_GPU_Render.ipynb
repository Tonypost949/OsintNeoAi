# 🎬 DJ DROP — Universal Intro Cloud GPU Renderer (NVIDIA T4 / A100)
**Renders all 600 frames in ~6–8 minutes on free Google Colab GPU and downloads DJDROP_intro.mp4.**

```python
# @title 🚀 1-Click Render (Click Play Button)
!nvidia-smi
!mkdir -p /content/render /content/frames
%cd /content/render

# 1. Download Blender Linux
!wget -q -nc https://download.blender.org/release/Blender5.2/blender-5.2.1-linux-x64.tar.xz || wget -q -nc https://download.blender.org/release/Blender4.3/blender-4.3.2-linux-x64.tar.xz
!tar -xf blender-*-linux-x64.tar.xz --strip-components=1 -C /content/render/

# 2. Download Packed Blend Asset
!wget -q -O universal_DJDROP.blend https://github.com/Tonypost949/OsintNeoAi/releases/download/render-assets/universal_DJDROP.blend

# 3. Configure CUDA / OptiX GPU Render Script & Install xvfb for EEVEE EGL Context
!apt-get update && apt-get install -y xvfb libgl1-mesa-glx libglu1-mesa
render_script = '''
import bpy
import os

out_dir = '/content/frames'
os.makedirs(out_dir, exist_ok=True)

# Enable CUDA / OptiX
prefs = bpy.context.preferences
cprefs = prefs.addons['cycles'].preferences
cprefs.get_devices()
for compute_type in ['OPTIX', 'CUDA']:
    try:
        cprefs.compute_device_type = compute_type
        for d in cprefs.devices:
            d.use = True
        print(f'Activated Cycles GPU compute: {compute_type}')
        break
    except Exception:
        pass

for s in bpy.data.scenes:
    s.render.filepath = os.path.join(out_dir, 'frame_')
    s.render.image_settings.file_format = 'PNG'
    s.render.image_settings.color_mode = 'RGBA'
    s.render.resolution_x = 1920
    s.render.resolution_y = 1080
    s.render.resolution_percentage = 100
    if hasattr(s, 'cycles'):
        s.cycles.device = 'GPU'
        s.cycles.samples = 32
        s.cycles.use_denoising = True
        s.cycles.denoiser = 'OPTIX'
'''
with open('/content/render/gpu_render.py', 'w') as f:
    f.write(render_script)

# 4. Render All 600 Frames on GPU (via xvfb to fix EGL EEVEE context errors)
!xvfb-run -a /content/render/blender -b universal_DJDROP.blend -P /content/render/gpu_render.py -s 1 -e 600 -a

# 5. Encode Mobile FastStart H.264 MP4
!ffmpeg -y -framerate 24 -i /content/frames/frame_%04d.png -c:v libx264 -preset slow -crf 18 -pix_fmt yuv420p -movflags +faststart /content/DJDROP_intro.mp4

# 6. Download to your device
from google.colab import files
files.download('/content/DJDROP_intro.mp4')
print('✅ ALL 600 FRAMES RENDERED & VIDEO DOWNLOADED!')
```
